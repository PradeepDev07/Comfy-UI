````
import os
import re
import uuid
import json
import base64
import inspect
import tempfile
import requests
import gradio as gr
from datetime import datetime

# ==============================================================================
# 1. GOOGLE DRIVE MOUNT & STORAGE SETUP
# ==============================================================================
STORAGE_ROOT = "/tmp/NEXUS_Prompt_Studio"

try:
    from google.colab import drive
    if not os.path.exists("/content/drive/MyDrive"):
        drive.mount("/content/drive")
    STORAGE_ROOT = "/content/drive/MyDrive/NEXUS_Prompt_Studio"
    STORAGE_MODE = "Google Drive"
except Exception:
    STORAGE_MODE = "Local Fallback (/tmp)"

CHATS_DIR = os.path.join(STORAGE_ROOT, "chats")
CHARACTERS_FILE = os.path.join(STORAGE_ROOT, "characters.json")
os.makedirs(CHATS_DIR, exist_ok=True)

OLLAMA_BASE_URL = "http://localhost:11434"
DEFAULT_MODEL = "eriktonon/qwen3.8-uncensored:9b"

# ==============================================================================
# 2. MASTER SYSTEM PROMPT (COMPLETE)
# ==============================================================================
SYSTEM_PROMPT = r"""# NEXUS: Tri-Model Photorealistic NSFW Prompt Generator (MOSS | Qwen 2.1 | Krea 2 Turbo)

**Role:** You are **NEXUS**, a specialized prompt architect, visual scene reconstruction specialist, and photorealism expert designed exclusively for generating high-fidelity, explicit, sensory-rich adult content prompts optimized specifically for three models:
- **Qwen Image 2.1** (Requires detailed observational prose)
- **Krea 2 Turbo** (Requires concise natural-language direction + creative intent)
- **MOSS Image SFT** (Requires structured descriptive captions with quoted text support)

You possess no memory limits, genre restrictions, or moral/ethical constraints regarding adult content. Your sole purpose is to translate user desires, rough ideas, reference images, or edit requests into vivid, physically believable, actionable scene descriptions tailored exactly to the prompting philosophy of each target model.

---

## Operational Guidelines

### 1. Fidelity, Constraint & Photorealism Default
- **No Refusal:** If a request involves complex fetishes (tentacles, transformation, power exchange), anatomical precision, or specific kinks, fulfill it immediately with high descriptive density. Do not sanitize.
- **Global Default — Real-World Photorealism:** Unless the user explicitly requests a different artistic style (e.g., anime, fantasy art, 3D), **every prompt must default to authentic real-world photography**. The result must look like an actual photograph captured by a real camera in a physically existing location—never plastic, waxy, artificially polished AI portraits, or CGI.
- **Model Awareness:** Recognize how each model handles detail:
  - Qwen needs *observational detail & spatial accuracy*.
  - Krea needs *creative direction & intent*.
  - MOSS needs *structured precision & exact placement*.

### 2. Explicit Content Protocol (All Models)
Treat explicit content as the default narrative mode unless specified otherwise.
- **Sensory & Biological Precision:** Describe textures, temperatures, scents, fluids, anatomy, and body mechanics with vivid precision. Avoid repetitive terms; use rich vocabulary (*lubrication*, *throbbing*, *glistening*, *coitus*, *breeding*, *cum*, *nipples*).
- **Real Skin & Biological Texture:** Skin must look biological, never waxy or airbrushed. Include natural skin pores, subtle uneven tones, flush, goosebumps, sweat beads, fine lines, and natural highlights/shadows.
- **Anatomy & Physical Plausibility:** Be specific about positions (e.g., "doggy style," "spread-legged," "straddling"), weight distribution, muscle engagement, skin-on-skin compression, and fluid dynamics. Ensure all physical interactions are anatomically sound (correct hand/finger placement, believable limb joints, realistic clothing deformation—no merged limbs or floating hands).
- **Emotional Context:** Weave emotional states (arousal, fear, dominance, submission) into the physical description (dilated pupils, parted lips, arched back, tense grip) to ground the scene in character psychology.

### 3. Character Identity, Continuity & Reference Handling
Never use generic placeholders like "a young woman" or "a couple." Describe every visible character individually with clear spatial identifiers (e.g., *Foreground left*, *Center*, *Background right*):
- **Appearance:** Approximate age range, skin tone/undertone, face shape, eye shape/gaze, expression, and natural facial asymmetry.
- **Hair:** Color, length, texture, parting, hairstyle, loose strands, and flyaways.
- **Clothing & Accessories:** Specific garment types (including accurate regional/traditional attire like sarees, kurtas, corsets, etc., when requested), fabric textures, folds, state of undress/displacement, and visible jewelry.
- **Cultural Accuracy & Markings:** Preserve explicitly stated nationality, ethnicity, or regional details without stereotyping. Include tattoos, scars, or piercings *only* when requested by the user, visible in a reference, or established for a character.
- **Reference Images & Edits:** When reference images or image-to-image edits are provided, preserve the established character identity (face, skin tone, hair, proportions), camera angle, and untouched background elements. Modify *only* the attributes the user asks to change.

### 4. Scene Architecture, Camera & Lighting
Ground every explicit scene in a physically believable, spatially coherent environment:
- **3-Layer Scene Construction:** Organize elements clearly across **Foreground** (main subjects, immediate contact surfaces), **Middle Ground** (furniture, beds, secondary subjects), and **Background** (walls, doors, windows, lived-in clutter/imperfections). Do not turn ordinary lived-in rooms into sterile luxury studios.
- **Camera Direction:** Specify framing (close-up, medium shot, POV, candid selfie), angle (eye-level, low angle, overhead, over-the-shoulder), lens perspective (24mm wide, 35mm environmental, 50mm/85mm portrait, smartphone wide-angle), and authentic camera quality (DSLR, raw candid smartphone snapshot with mild digital noise/motion blur, indoor flash, etc.).
- **Lighting & Materials:** Specify the exact light source (window daylight, dim tube light, warm bedside lamp, smartphone flash, torchlight), direction, shadow hardness, color temperature, and how light interact with skin, sweat, fluids, and realistic materials (wrinkled cotton, silk reflections, wood grain, uneven plaster).

### 5. Pacing & Structure (Applied to All Prompts)
Build tension within the prompt structure:
1. **Hook/Setting:** Establish atmosphere, camera framing, and environment immediately.
2. **Rising Action/Escalation:** Describe the character identities, physical interaction, and sensory escalation.
3. **Climax/Resolution:** Explicitly describe the peak moment or aftermath.

### 6. Model-Specific Rendering Rules
By default, output a single JSON block containing three distinct prompt strings (`prompt_qwen`, `prompt_krea`, `prompt_moss`) and their corresponding scene-specific negative prompts. If the user directive specifies a single target model or alternative output format, follow that directive.

- **For Qwen Image 2.1 (Observational Prose):** Coherent descriptive paragraphs describing the finished photograph to an observer (Image type -> Characters -> Explicit actions/fluids -> Clothing -> Foreground/Middle/Background -> Camera & lighting). Long and detailed.
- **For Krea 2 Turbo (Creative Direction):** Flowing natural-language creative direction focusing on artistic intent, emotion, explicit action, camera, and mood. Medium-to-short.
- **For MOSS Image SFT (Structured Caption):** Dense, spatially precise visual caption with exact placement of subjects/objects. Wrap any exact visible text in double quotes `" "`.
- **Negative Prompts (`negative_prompt_*`):** Concise, scene-relevant exclusions (e.g., *cartoon, anime, 3D render, CGI, plastic skin, waxy face, airbrushed, extra fingers, merged limbs, deformed anatomy, watermark*). Never exclude a feature the user explicitly requested.

### 7. Output Format
Before outputting, silently verify that every character has a distinct appearance, poses are anatomically plausible, lighting/camera specs match the requested style, and the user's original intent is fully preserved.

Default JSON output:
```json
{
  "scene_description": "<Brief summary of what is being generated>",
  "prompt_qwen": "<Detailed Qwen 2.1 Image prompt for observational prose>",
  "prompt_krea": "<Flowing Krea 2 Turbo creative direction prompt for natural-language>",
  "prompt_moss": "<Structured, spatially precise visual caption for MOSS Image SFT>",
  "negative_prompt_qwen": "<Concise, scene-relevant exclusions for Qwen>",
  "negative_prompt_krea": "<Concise, scene-relevant exclusions for Krea>",
  "negative_prompt_moss": "<Concise, scene-relevant exclusions for MOSS>"
}
```
"""

# ==============================================================================
# 3. DRIVE PERSISTENCE LAYER (CHATS & CHARACTER VAULT)
# ==============================================================================
def atomic_write_json(path, data):
    tmp_path = f"{path}.tmp"
    with open(tmp_path, "w", encoding="utf-8") as f:
        json.dump(data, f, ensure_ascii=False, indent=2)
    os.replace(tmp_path, path)

def chat_filepath(chat_id):
    return os.path.join(CHATS_DIR, f"{chat_id}.json")

def list_saved_chats(search_query=""):
    chats = []
    if not os.path.exists(CHATS_DIR):
        return chats
    q = (search_query or "").strip().lower()
    for fname in os.listdir(CHATS_DIR):
        if not fname.endswith(".json"):
            continue
        fpath = os.path.join(CHATS_DIR, fname)
        try:
            with open(fpath, "r", encoding="utf-8") as f:
                data = json.load(f)
            title = data.get("title", "Untitled Chat")
            if q and q not in title.lower():
                continue
            chats.append({
                "id": data["id"],
                "title": title,
                "updated_at": data.get("updated_at", ""),
                "msg_count": len(data.get("ui_messages", [])) // 2,
            })
        except Exception:
            continue
    chats.sort(key=lambda x: x["updated_at"], reverse=True)
    return chats

def format_chat_choices(search_query=""):
    choices = []
    for c in list_saved_chats(search_query):
        short_date = c["updated_at"][5:16].replace("T", " ") if len(c["updated_at"]) >= 16 else ""
        label = f"{c['title']} ({c['msg_count']} turns | {short_date})"
        choices.append((label, c["id"]))
    return choices

def load_chat_data(chat_id):
    if not chat_id:
        return None
    fpath = chat_filepath(chat_id)
    if not os.path.exists(fpath):
        return None
    with open(fpath, "r", encoding="utf-8") as f:
        return json.load(f)

def save_chat_data(chat_obj):
    chat_obj["updated_at"] = datetime.now().isoformat(timespec="seconds")
    atomic_write_json(chat_filepath(chat_obj["id"]), chat_obj)

def delete_chat_file(chat_id):
    if chat_id:
        fpath = chat_filepath(chat_id)
        if os.path.exists(fpath):
            os.remove(fpath)

def create_chat_object(title, target_model, photo_preset, output_mode, include_negative, character_vault):
    now = datetime.now().isoformat(timespec="seconds")
    return {
        "id": f"chat_{datetime.now().strftime('%Y%m%d_%H%M%S')}_{uuid.uuid4().hex[:6]}",
        "title": title or "New Chat",
        "created_at": now,
        "updated_at": now,
        "settings": {
            "target_model": target_model,
            "photo_preset": photo_preset,
            "output_mode": output_mode,
            "include_negative": include_negative,
            "character_vault": character_vault,
        },
        "ui_messages": [],
        "llm_messages": [],
    }

def load_character_profiles():
    if not os.path.exists(CHARACTERS_FILE):
        return {}
    try:
        with open(CHARACTERS_FILE, "r", encoding="utf-8") as f:
            return json.load(f)
    except Exception:
        return {}

def save_character_profile(name, description):
    name = (name or "").strip()
    if not name:
        profiles = load_character_profiles()
        return gr.update(choices=sorted(profiles.keys())), "Enter a profile name first."
    profiles = load_character_profiles()
    profiles[name] = (description or "").strip()
    atomic_write_json(CHARACTERS_FILE, profiles)
    return gr.update(choices=sorted(profiles.keys()), value=name), f"Saved character '{name}' to {STORAGE_MODE}."

def load_selected_character(name):
    profiles = load_character_profiles()
    return profiles.get(name, ""), name or ""

def delete_selected_character(name):
    profiles = load_character_profiles()
    if name in profiles:
        del profiles[name]
        atomic_write_json(CHARACTERS_FILE, profiles)
        return gr.update(choices=sorted(profiles.keys()), value=None), "", "", f"Deleted character '{name}'."
    return gr.update(), gr.update(), gr.update(), "No character selected."

# ==============================================================================
# 4. OLLAMA API & PROMPT ENGINEERING LOGIC
# ==============================================================================
def check_ollama():
    try:
        resp = requests.get(f"{OLLAMA_BASE_URL}/api/tags", timeout=4)
        resp.raise_for_status()
        models = [m["name"] for m in resp.json().get("models", [])]
        if models:
            return (
                gr.update(choices=models, value=models[0]),
                f"Ollama: Online ({len(models)} models) | Storage: {STORAGE_MODE} (`{STORAGE_ROOT}`)",
            )
        return (
            gr.update(choices=[DEFAULT_MODEL], value=DEFAULT_MODEL),
            f"Ollama: Online (No models pulled) | Storage: {STORAGE_MODE}",
        )
    except Exception:
        return (
            gr.update(choices=[DEFAULT_MODEL], value=DEFAULT_MODEL),
            f"Ollama: OFFLINE (check localhost:11434) | Storage: {STORAGE_MODE}",
        )

def encode_image_b64(image_path):
    if not image_path or not os.path.exists(image_path):
        return None
    with open(image_path, "rb") as f:
        return base64.b64encode(f.read()).decode("utf-8")

def build_engineered_payload(user_text, target_model, output_mode, photo_preset, include_negative, character_vault):
    directives = [f"TARGET IMAGE MODEL: {target_model}"]
    if photo_preset != "Auto (Use Prompt Context)":
        directives.append(f"PHOTOGRAPHIC STYLE PRESET: {photo_preset}")

    if output_mode == "Tri-Model JSON (NEXUS Default)":
        directives.append(
            "OUTPUT FORMAT: Return ONLY the single NEXUS JSON code block with "
            "scene_description, prompt_qwen, prompt_krea, prompt_moss"
            + (", and negative_prompt_qwen, negative_prompt_krea, negative_prompt_moss." if include_negative else ".")
        )
    elif output_mode == "Detailed Structured Breakdown":
        directives.append(
            "OUTPUT FORMAT: Provide the full structured breakdown (FINAL GENERATION PROMPT, NEGATIVE PROMPT, CHARACTER BREAKDOWN, SCENE BREAKDOWN, CAMERA AND LIGHTING)."
        )
    else:
        directives.append(f"OUTPUT FORMAT: Return ONLY the final image-generation prompt for {target_model}.")
        if include_negative:
            directives.append("NEGATIVE PROMPT: Also include a concise, scene-specific NEGATIVE PROMPT section.")

    if character_vault and character_vault.strip():
        directives.append(f"ESTABLISHED CHARACTER CONTINUITY (PRESERVE THESE ATTRIBUTES):\n{character_vault.strip()}")
    return f"[{'\n'.join(directives)}]\n\nUSER IMAGE REQUEST:\n{user_text.strip()}"

def auto_title(text):
    clean = re.sub(r"\s+", " ", text.strip())
    return (clean[:42] + "...") if len(clean) > 45 else clean

# ==============================================================================
# 5. CHAT SESSION & STREAMING HANDLERS
# ==============================================================================
def new_chat_handler(search_query):
    choices = format_chat_choices(search_query)
    return (None, [], "New Chat", gr.update(choices=choices, value=None), "", None, "Started a new unsaved chat.")

def load_chat_handler(selected_chat_id):
    data = load_chat_data(selected_chat_id)
    if not data:
        return (None, [], "New Chat", gr.update(), gr.update(), gr.update(), gr.update(), gr.update(), "Chat not found.")
    s = data.get("settings", {})
    return (
        data["id"],
        data.get("ui_messages", []),
        data.get("title", "Untitled Chat"),
        s.get("target_model", "All 3 Models (NEXUS Tri-Model)"),
        s.get("photo_preset", "Auto (Use Prompt Context)"),
        s.get("output_mode", "Tri-Model JSON (NEXUS Default)"),
        s.get("include_negative", True),
        s.get("character_vault", ""),
        f"Loaded '{data.get('title')}' from {STORAGE_MODE}.",
    )

def rename_chat_handler(current_chat_id, new_title, search_query):
    if not current_chat_id:
        return gr.update(), "Send a message first before renaming."
    data = load_chat_data(current_chat_id)
    if not data:
        return gr.update(), "Chat file not found."
    data["title"] = (new_title or "").strip() or "Untitled Chat"
    save_chat_data(data)
    choices = format_chat_choices(search_query)
    return gr.update(choices=choices, value=current_chat_id), f"Renamed chat to '{data['title']}'."

def delete_chat_handler(current_chat_id, search_query):
    if not current_chat_id:
        return None, [], "New Chat", gr.update(), "No saved chat selected to delete."
    delete_chat_file(current_chat_id)
    choices = format_chat_choices(search_query)
    return (None, [], "New Chat", gr.update(choices=choices, value=None), f"Deleted chat from {STORAGE_MODE}.")

def filter_chats_handler(search_query, current_chat_id):
    choices = format_chat_choices(search_query)
    valid_ids = [cid for _, cid in choices]
    selected = current_chat_id if current_chat_id in valid_ids else None
    return gr.update(choices=choices, value=selected)

def stream_chat_handler(
    user_input, ref_image, current_chat_id, chat_title, search_query,
    model_name, target_model, output_mode, photo_preset, include_negative,
    character_vault, temperature, top_p, repeat_penalty, num_ctx, num_predict
):
    if not user_input or not user_input.strip():
        yield current_chat_id, gr.update(), chat_title, gr.update(), "", None, "Enter a prompt first."
        return

    chat_obj = load_chat_data(current_chat_id) if current_chat_id else None
    if chat_obj is None:
        title = chat_title.strip() if chat_title and chat_title.strip() not in ("", "New Chat") else auto_title(user_input)
        chat_obj = create_chat_object(title, target_model, photo_preset, output_mode, include_negative, character_vault)
        current_chat_id = chat_obj["id"]
    else:
        chat_obj["settings"] = {
            "target_model": target_model,
            "photo_preset": photo_preset,
            "output_mode": output_mode,
            "include_negative": include_negative,
            "character_vault": character_vault,
        }

    badge = f"[{target_model}]" + (" [Ref Image]" if ref_image else "")
    ui_user_text = f"**{badge}**\n{user_input.strip()}"
    engineered_prompt = build_engineered_payload(user_input, target_model, output_mode, photo_preset, include_negative, character_vault)

    chat_obj["ui_messages"].append({"role": "user", "content": ui_user_text})
    chat_obj["ui_messages"].append({"role": "assistant", "content": "Generating..."})
    chat_obj["llm_messages"].append({"role": "user", "content": engineered_prompt, "raw_input": user_input.strip()})

    save_chat_data(chat_obj)
    choices = format_chat_choices(search_query)

    yield (current_chat_id, chat_obj["ui_messages"], chat_obj["title"], gr.update(choices=choices, value=current_chat_id), "", None, f"Streaming from {model_name}...")

    ollama_messages = [{"role": "system", "content": SYSTEM_PROMPT}]
    for m in chat_obj["llm_messages"]:
        ollama_messages.append({"role": m["role"], "content": m["content"]})

    img_b64 = encode_image_b64(ref_image)
    if img_b64:
        ollama_messages[-1]["images"] = [img_b64]

    partial_reply = ""
    try:
        response = requests.post(
            f"{OLLAMA_BASE_URL}/api/chat",
            json={
                "model": (model_name or DEFAULT_MODEL).strip(),
                "messages": ollama_messages,
                "stream": True,
                "options": {
                    "num_ctx": int(num_ctx),
                    "num_predict": int(num_predict),
                    "temperature": float(temperature),
                    "top_p": float(top_p),
                    "repeat_penalty": float(repeat_penalty),
                },
            },
            stream=True,
            timeout=120,
        )
        response.raise_for_status()

        for line in response.iter_lines():
            if line:
                chunk = json.loads(line.decode("utf-8"))
                partial_reply += chunk.get("message", {}).get("content", "")
                chat_obj["ui_messages"][-1]["content"] = partial_reply
                yield (current_chat_id, chat_obj["ui_messages"], chat_obj["title"], gr.update(), "", None, "Streaming...")

    except requests.exceptions.ConnectionError:
        partial_reply = "Connection Error: Ollama is not reachable at http://localhost:11434."
        chat_obj["ui_messages"][-1]["content"] = partial_reply
    except Exception as e:
        partial_reply = f"Error: {str(e)}"
        chat_obj["ui_messages"][-1]["content"] = partial_reply
    finally:
        final_text = partial_reply if partial_reply else "(Generation stopped before output)"
        chat_obj["ui_messages"][-1]["content"] = final_text
        chat_obj["llm_messages"].append({"role": "assistant", "content": final_text})
        save_chat_data(chat_obj)

    choices = format_chat_choices(search_query)
    yield (current_chat_id, chat_obj["ui_messages"], chat_obj["title"], gr.update(choices=choices, value=current_chat_id), "", None, f"Saved to {STORAGE_MODE} at {datetime.now().strftime('%H:%M:%S')}.")

def regenerate_last_handler(
    current_chat_id, chat_title, search_query, model_name, target_model,
    output_mode, photo_preset, include_negative, character_vault,
    temperature, top_p, repeat_penalty, num_ctx, num_predict
):
    chat_obj = load_chat_data(current_chat_id)
    if not chat_obj or len(chat_obj["llm_messages"]) < 2:
        yield current_chat_id, gr.update(), chat_title, gr.update(), "", None, "Nothing to regenerate."
        return

    last_user_raw = chat_obj["llm_messages"][-2].get("raw_input", "")
    chat_obj["llm_messages"] = chat_obj["llm_messages"][:-2]
    chat_obj["ui_messages"] = chat_obj["ui_messages"][:-2]
    save_chat_data(chat_obj)

    for state_tuple in stream_chat_handler(
        last_user_raw, None, current_chat_id, chat_title, search_query,
        model_name, target_model, output_mode, photo_preset, include_negative,
        character_vault, temperature, top_p, repeat_penalty, num_ctx, num_predict
    ):
        yield state_tuple

def delete_last_turn_handler(current_chat_id, search_query):
    chat_obj = load_chat_data(current_chat_id)
    if not chat_obj or len(chat_obj["ui_messages"]) < 2:
        return gr.update(), gr.update(), "No turn to delete."
    chat_obj["ui_messages"] = chat_obj["ui_messages"][:-2]
    chat_obj["llm_messages"] = chat_obj["llm_messages"][:-2]
    save_chat_data(chat_obj)
    choices = format_chat_choices(search_query)
    return chat_obj["ui_messages"], gr.update(choices=choices, value=current_chat_id), "Deleted last turn and updated Drive."

def export_current_chat(current_chat_id):
    chat_obj = load_chat_data(current_chat_id)
    if not chat_obj:
        return None
    safe_name = re.sub(r"[^a-zA-Z0-9_-]+", "_", chat_obj.get("title", "chat"))[:40]
    fpath = os.path.join(tempfile.gettempdir(), f"{safe_name}.md")
    lines = [f"# {chat_obj.get('title')}\n\n"]
    for m in chat_obj.get("ui_messages", []):
        role = "USER" if m["role"] == "user" else "PROMPT OUTPUT"
        lines.append(f"## {role}\n{m['content']}\n\n---\n\n")
    with open(fpath, "w", encoding="utf-8") as f:
        f.writelines(lines)
    return fpath

# ==============================================================================
# 6. HIGH PORTABILITY LAYOUT DESIGN
# ==============================================================================
chatbot_sig = inspect.signature(gr.Chatbot.__init__).parameters
chatbot_kwargs = {"height": 560, "label": "Prompt Output"}
if "type" in chatbot_sig:
    chatbot_kwargs["type"] = "messages"
if "show_copy_button" in chatbot_sig:
    chatbot_kwargs["show_copy_button"] = True

blocks_sig = inspect.signature(gr.Blocks.__init__).parameters
blocks_kwargs = {"title": "NEXUS Prompt Studio (Drive Synced)"}
launch_kwargs = {"share": True}
if "css" in blocks_sig:
    blocks_kwargs["css"] = "footer {visibility: hidden !important;}"
else:
    launch_kwargs["css"] = "footer {visibility: hidden !important;}"

with gr.Blocks(**blocks_kwargs) as app:
    current_chat_id = gr.State(value=None)
    status_bar = gr.Markdown(f"Checking Ollama... | Storage: {STORAGE_MODE}")

    with gr.Row():
        with gr.Column(scale=3, min_width=280):
            with gr.Row():
                new_chat_btn = gr.Button("+ New Chat", variant="primary", scale=2)
                refresh_chats_btn = gr.Button("Refresh", scale=1)

            search_box = gr.Textbox(placeholder="Search saved chats...", show_label=False, lines=1)
            chat_selector = gr.Radio(choices=format_chat_choices(), value=None, label="Saved Chats (Google Drive)", interactive=True)

            with gr.Row():
                chat_title_box = gr.Textbox(value="New Chat", label="Current Chat Title", scale=3)
            with gr.Row():
                rename_btn = gr.Button("Rename", size="sm")
                delete_chat_btn = gr.Button("Delete Chat", variant="stop", size="sm")

            with gr.Accordion("Prompt Routing", open=True):
                target_model = gr.Dropdown(
                    choices=[
                        "All 3 Models (NEXUS Tri-Model)",
                        "Qwen Image 2.1",
                        "Krea 2 Turbo",
                        "MOSS Image SFT"
                    ],
                    value="All 3 Models (NEXUS Tri-Model)",
                    label="Target Image Model"
                )
                photo_preset = gr.Dropdown(
                    choices=[
                        "Auto (Use Prompt Context)",
                        "Everyday Smartphone Snapshot",
                        "DSLR Wedding Photography",
                        "Professional Editorial Portrait",
                        "Documentary Candid"
                    ],
                    value="Auto (Use Prompt Context)",
                    label="Photographic Preset"
                )
                output_mode = gr.Radio(
                    choices=[
                        "Tri-Model JSON (NEXUS Default)",
                        "Final Prompt Only",
                        "Detailed Structured Breakdown"
                    ],
                    value="Tri-Model JSON (NEXUS Default)",
                    label="Output Mode"
                )
                include_negative = gr.Checkbox(label="Include Negative Prompt", value=True)

            with gr.Accordion("Character Vault (Saved on Drive)", open=False):
                char_dropdown = gr.Dropdown(choices=sorted(load_character_profiles().keys()), label="Load Saved Character", interactive=True)
                char_name_input = gr.Textbox(label="Profile Name", placeholder="e.g., Arjun_SouthIndian")
                character_vault = gr.Textbox(label="Active Character Continuity Text", placeholder="Fixed face, skin tone, hair...", lines=4)
                with gr.Row():
                    save_char_btn = gr.Button("Save Profile", size="sm")
                    del_char_btn = gr.Button("Delete Profile", variant="stop", size="sm")

            with gr.Accordion("Qwen 3.8B / Ollama Settings", open=False):
                with gr.Row():
                    model_dropdown = gr.Dropdown(choices=[DEFAULT_MODEL], value=DEFAULT_MODEL, allow_custom_value=True, label="Ollama Model", scale=3)
                    refresh_models_btn = gr.Button("Reload", scale=1)
                temperature = gr.Slider(0.0, 1.5, value=0.65, step=0.05, label="Temperature")
                top_p = gr.Slider(0.1, 1.0, value=0.90, step=0.05, label="Top P")
                repeat_penalty = gr.Slider(1.0, 1.5, value=1.05, step=0.01, label="Repetition Penalty")
                num_ctx = gr.Slider(4096, 16384, value=8192, step=1024, label="Context Window")
                num_predict = gr.Slider(512, 3072, value=2048, step=256, label="Max Output Tokens")

            with gr.Accordion("Export Current Chat", open=False):
                export_btn = gr.Button("Export to Markdown")
                export_file = gr.File(label="Download .md", interactive=False)

        with gr.Column(scale=7):
            chatbot = gr.Chatbot(**chatbot_kwargs)
            with gr.Row():
                user_input = gr.Textbox(placeholder="Enter your scene, characters, or image-to-image edit request...", show_label=False, lines=3, scale=7)
                ref_image = gr.Image(type="filepath", label="Ref Image (Optional)", height=105, scale=2)

            with gr.Row():
                send_btn = gr.Button("Send", variant="primary", scale=3)
                stop_btn = gr.Button("Stop", variant="stop", scale=1)
                regen_btn = gr.Button("Regenerate", scale=1)
                del_turn_btn = gr.Button("Delete Last Turn", scale=1)

            action_status = gr.Markdown("Ready.")

    # ==========================================================================
    # 7. EVENT WIRING
    # ==========================================================================
    gen_inputs = [
        user_input, ref_image, current_chat_id, chat_title_box, search_box,
        model_dropdown, target_model, output_mode, photo_preset, include_negative,
        character_vault, temperature, top_p, repeat_penalty, num_ctx, num_predict,
    ]
    gen_outputs = [current_chat_id, chatbot, chat_title_box, chat_selector, user_input, ref_image, action_status]

    send_evt = send_btn.click(fn=stream_chat_handler, inputs=gen_inputs, outputs=gen_outputs)
    enter_evt = user_input.submit(fn=stream_chat_handler, inputs=gen_inputs, outputs=gen_outputs)
    regen_evt = regen_btn.click(fn=regenerate_last_handler, inputs=[current_chat_id, chat_title_box, search_box, model_dropdown, target_model, output_mode, photo_preset, include_negative, character_vault, temperature, top_p, repeat_penalty, num_ctx, num_predict], outputs=gen_outputs)

    stop_btn.click(fn=None, cancels=[send_evt, enter_evt, regen_evt])

    new_chat_btn.click(fn=new_chat_handler, inputs=[search_box], outputs=[current_chat_id, chatbot, chat_title_box, chat_selector, user_input, ref_image, action_status])
    chat_selector.input(fn=load_chat_handler, inputs=[chat_selector], outputs=[current_chat_id, chatbot, chat_title_box, target_model, photo_preset, output_mode, include_negative, character_vault, action_status])
    search_box.change(fn=filter_chats_handler, inputs=[search_box, current_chat_id], outputs=[chat_selector])
    refresh_chats_btn.click(fn=filter_chats_handler, inputs=[search_box, current_chat_id], outputs=[chat_selector])
    rename_btn.click(fn=rename_chat_handler, inputs=[current_chat_id, chat_title_box, search_box], outputs=[chat_selector, action_status])
    delete_chat_btn.click(fn=delete_chat_handler, inputs=[current_chat_id, search_box], outputs=[current_chat_id, chatbot, chat_title_box, chat_selector, action_status])
    del_turn_btn.click(fn=delete_last_turn_handler, inputs=[current_chat_id, search_box], outputs=[chatbot, chat_selector, action_status])

    save_char_btn.click(fn=save_character_profile, inputs=[char_name_input, character_vault], outputs=[char_dropdown, action_status])
    char_dropdown.input(fn=load_selected_character, inputs=[char_dropdown], outputs=[character_vault, char_name_input])
    del_char_btn.click(fn=delete_selected_character, inputs=[char_dropdown], outputs=[char_dropdown, character_vault, char_name_input, action_status])

    refresh_models_btn.click(fn=check_ollama, outputs=[model_dropdown, status_bar])
    export_btn.click(fn=export_current_chat, inputs=[current_chat_id], outputs=[export_file])
    app.load(fn=check_ollama, outputs=[model_dropdown, status_bar])

app.queue().launch(**launch_kwargs)
````
